### 1.1 — Listas y operaciones básicas

Tienes la siguiente lista con la esperanza de vida de 10 países (en años).

````python
# Lista dada — NO modificar
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]
````

**Ejercicio 1.1** — Usando Python puro (sin Pandas ni funciones built-in como `sum()` o `max()`):
- Calcula el **promedio** de la lista usando un bucle `for`
- Encuentra el **valor máximo** usando un bucle `for`
- Encuentra el **valor mínimo** usando un bucle `for`

````python
# Tu código aquí
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]
total_suma= 0
#---codigo Promedio
for años_de_vida in esperanzas_de_vida:
    total_suma += años_de_vida
promedio = total_suma / len (esperanzas_de_vida)
#---codigo max
max_edad = esperanzas_de_vida[0]
for años_de_vida in esperanzas_de_vida:
    if años_de_vida > max_edad:
        max_edad = años_de_vida
#---codigo min
min_edad = esperanzas_de_vida[0]
for años_de_vida in esperanzas_de_vida:
    if años_de_vida < min_edad:
        min_edad = años_de_vida
# Promedio

print (f"Promedio de vida {promedio:.2f}")
# Máximo

print(f"Valor máximo: {max_edad}")
# Mínimo

print(f"Valor mínimo: {min_edad}")

````

### 1.3 — Funciones

**Ejercicio 1.3** — Crea una función llamada `clasificar_esperanza_vida` que reciba un número (años) y retorne:
- `"Alta"` si es >= 75
- `"Media"` si es entre 60 y 74.9
- `"Baja"` si es < 60

Prueba la función con al menos 5 valores distintos e imprime los resultados.

````python
# Tu código aquí
esperanzas_de_vida = [72.3, 81.5, 60.1, 78.9, 55.4, 83.2, 67.8, 74.0, 49.3, 79.6]

def clasificar_esperanza_vida(edad):
    if edad >= 75:
        return "Alta"
    elif 60 <= edad < 75:
        return "Media"
    else:
        return "Baja"

# Pruebas
for valor in esperanzas_de_vida:
    print(valor, clasificar_esperanza_vida(valor))
````

### 2.2 — Limpieza de datos

🔧 **Ejercicio 2.2** — El dataset tiene algunos valores nulos (¡datos reales siempre tienen este problema!).  
1. Rellena los valores nulos de columnas **numéricas** con la **media** de cada columna
2. Verifica que ya no queden nulos
3. Explica en una celda de texto: ¿Por qué usamos la media y no simplemente eliminamos esas filas?

````python
# Tu código aquí

#--- 2.2.1  Rellenar nulos en columnas numéricas con la media de cada columna

sm_nuevo = sm.copy()

num_cols = sm_nuevo.select_dtypes(include='number').columns
medias = sm_nuevo[num_cols].mean()

sm_nuevo[num_cols] = sm_nuevo[num_cols].fillna(medias)

print("Valores nulos por columna después del fillna:")
print(sm_nuevo.isnull().sum())

#--- 2.2.2 Verificación rápida
assert sm_nuevo[num_cols].isnull().sum().sum() == 0, "Quedan valores nulos en columnas numéricas"
````

**Responde aquí** (doble clic para editar):

*¿Por qué usamos la media para rellenar nulos en vez de eliminar las filas?*

> Tu respuesta: 

Al analizar el dataset, encontré que la cantidad de valores nulos era mínima en comparación con la información total. Específicamente, solo había 10 datos máximos, faltantes por cada columna con nulos frente a un total de 159 filas con datos importantes.

Decidí aplicar la media porque, al tener una base de datos tan sólida, el promedio resulta muy confiable y nos permite conservar casi toda la información. Si hubiera optado por borrar , habría desperdiciado 149 datos reales que son muy valiosos para el análisis. En este caso, completar los huecos es mucho más "rentable" que eliminar columnas o registros.

### 2.3 — Estadísticas descriptivas

🔧 **Ejercicio 2.3**
1. Muestra el resumen estadístico completo del dataset
2. ¿Cuál es la esperanza de vida **promedio** mundial?
3. ¿Cuál es el país con **mayor** gasto en salud? ¿Y el de **menor**?
4. ¿Cuál es la **mediana** de la mortalidad infantil?

````python
# Tu código aquí
# 1. Resumen estadístico completo
print("Resumen estadístico del dataset:")
print(sm.describe())

# 2. Esperanza de vida promedio mundial
esperanza_vida_promedio = sm['esperanza_vida'].mean()
print(f"\nEsperanza de vida promedio mundial: {esperanza_vida_promedio:.2f} años")

# 3. País con mayor y menor gasto en salud
pais_max_gasto = sm.loc[sm['gasto_salud_usd'].idxmax()]
pais_min_gasto = sm.loc[sm['gasto_salud_usd'].idxmin()]
print(f"\nPaís con mayor gasto en salud: {pais_max_gasto['pais']} (${pais_max_gasto['gasto_salud_usd']:.2f})")
print(f"País con menor gasto en salud: {pais_min_gasto['pais']} (${pais_min_gasto['gasto_salud_usd']:.2f})")

# 4. Mediana de la mortalidad infantil
mediana_mortalidad = sm['mortalidad_infantil'].median()
print(f"\nMediana de la mortalidad infantil: {mediana_mortalidad:.2f} muertes por 1,000 nacidos vivos")
````

### 2.5 — Agrupaciones

🔧 **Ejercicio 2.5** — Usa `groupby()` para:
1. Calcular la **esperanza de vida promedio** por `region`
2. Calcular el **gasto en salud promedio** por `nivel_ingresos`
3. Calcular el **número de países** por región
4. ¿Qué región tiene la mayor esperanza de vida promedio? ¿Y la menor?

````python
# Tu código aquí
# 1. Esperanza de vida promedio por región
esperanza_por_region = sm.groupby('region')['esperanza_vida'].mean()
print("Esperanza de vida promedio por región:")
print(esperanza_por_region)

# 2. Gasto en salud promedio por nivel_ingresos
gasto_por_ingresos = sm.groupby('nivel_ingresos')['gasto_salud_usd'].mean()
print("\nGasto en salud promedio por nivel_ingresos:")
print(gasto_por_ingresos)

# 3. Número de países por región
paises_por_region = sm.groupby('region')['pais'].count()
print("\nNúmero de países por región:")
print(paises_por_region)

# 4. Región con mayor y menor esperanza de vida promedio
region_max_esperanza = esperanza_por_region.idxmax()
region_min_esperanza = esperanza_por_region.idxmin()
print(f"\nRegión con mayor esperanza de vida promedio: {region_max_esperanza} ({esperanza_por_region.max():.2f})")
print(f"Región con menor esperanza de vida promedio: {region_min_esperanza} ({esperanza_por_region.min():.2f})")
````

### 2.6 — Nueva columna calculada

🔧 **Ejercicio 2.6**
1. Crea una nueva columna llamada `categoria_vida` aplicando la función `clasificar_esperanza_vida` que creaste en la Parte 1 a la columna `esperanza_vida`
2. ¿Cuántos países caen en cada categoría? (usa `value_counts()`)
3. ¿Qué porcentaje del total representa cada categoría?

````python
# Tu código aquí
# Crear columna usando la función ya definida (clasificar_esperanza_vida)
sm['categoria_vida'] = sm['esperanza_vida'].apply(lambda x: clasificar_esperanza_vida(x) if pd.notnull(x) else None)

# Conteo por categoría
conteos_categoria = sm['categoria_vida'].value_counts()
print("Conteos por categoría:")
print(conteos_categoria)

# Porcentaje por categoría sin numpy
porcentaje_categoria = (conteos_categoria / conteos_categoria.sum()) * 100
print("\nPorcentaje por categoría:")
print(porcentaje_categoria.round(2))
````
